## 3. Understand the kernel and execution order

The **notebook** stores cell text and saved outputs. The **kernel** runs
your code and holds variables in memory. Code cells in one notebook
normally share that memory.

Run the following three cells in order: **A**, **B**, then **C**.

````python
# Cell A
a = 10
````

````python
# Cell B
b = a + 1
````

````python
# Cell C
print(a, b)
````

The result is `10 11`. Now try this:

1. Change Cell A to `a = 20` and run **only A**.
2. Run **C**. It prints `20 11`: B has not run again, so `b` still holds its old value.
3. Run **B**, then **C**. The result is now `20 21`.

**Editing does not execute code.** Rerun a changed cell and every later
cell that depends on it. Existing outputs do not update automatically.

### Read the execution indicator

| Beside a Code cell | Meaning |
| --- | --- |
| `[ ]` | The cell has no execution count yet. |
| `[*]` | The cell is executing or waiting for the kernel. |
| `[7]` | This was the kernel's seventh numbered execution. |

The numbers record execution order, which can differ from page order.

### Restart and run all

Use **Kernel > Restart Kernel and Run All Cells…** and confirm. If your
menu uses a different label, restart the kernel, then use **Run > Run All Cells**.
The kernel forgets its variables and executes the notebook from the top.
This checks that the definitions you need are present and ordered correctly.

Restarting alone keeps your cell text. Old outputs can remain visible
even though their variables no longer exist in memory. For a calculation
that takes too long, try **Kernel > Interrupt Kernel**, then restart if
necessary. Avoid repeatedly starting the same slow cell.

The unfinished exercise cells later in this tutorial display a reminder;
they do not prevent a fresh **Run All** from reaching the end.

### Sample a sine curve

`np.linspace` creates equally spaced points, including both endpoints.
NumPy's trigonometric functions use **radians**.

````python
sample_count = 201  # Try 21 later, then restore 201.
x = np.linspace(0, np.pi, sample_count)
y = np.sin(x)
print(x.shape, y.shape)
````

For 201 points, the output is `(201,) (201,)`. The variables `x` and `y`
are available to the next cell.

### Approximate an integral with SciPy

The `trapezoid` function, imported from `scipy.integrate`, approximates
an integral from sampled data. The exact comparison is

$$\int_0^\pi \sin(x)\,dx = 2.$$

````python
area = trapezoid(y, x=x)
print(f"Integral estimate: {area:.6f}")
print(f"Absolute error: {abs(area - 2):.6f}")
````

With 201 sample points, expect:

```text
Integral estimate: 1.999959
Absolute error: 0.000041
```

The format `:.6f` displays six decimal places without changing the stored value.

**Try it:** change `sample_count` to `21` in the sampling cell, rerun it,
then rerun the integration cell. Compare the error. Restore `201` and
rerun both cells before continuing.

## 5. Draw and explain a figure

The next cell uses `x` and `y` from the sampling cell. Run it to display
a plot inside the notebook. Then change the title or `line_color` and
run it again. `plt.subplots()` creates a fresh figure for each run.

````python
line_color = "tab:blue"  # Try "tab:orange" or "tab:green".
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(x, y, color=line_color, label="sin(x)")
ax.fill_between(x, y, color=line_color, alpha=0.15)
ax.set_xlabel("x (radians)")
ax.set_ylabel("y")
ax.set_title("Area under a sine curve")
ax.grid(True, alpha=0.3)
ax.legend()
fig.tight_layout()
plt.show()
````

### When something goes wrong

| What you see | What to try |
| --- | --- |
| `FileNotFoundError` | Upload the file to the notebook's folder and check its exact name. Inspect the current folder with the cell above. |
| `NameError` | Check spelling and run the defining cell. For `np`, `plt` or `trapezoid`, rerun the imports. |
| `SyntaxError` or `IndentationError` | Check the highlighted line, brackets, quotes, colons and indentation. Fix the code and rerun. |
| `TypeError` or `ValueError` | Read the last line of the traceback. Check the inputs and array shapes. |
| A cell stays at `[*]` | Allow time for first loading. For a stuck calculation, interrupt the kernel, then restart if necessary. |
| Old or unexpected output | Rerun definitions and dependent cells, or restart and run all. |
| An import fails | Select a Python kernel, restart and rerun imports. Ask the instructor if it still fails. |
